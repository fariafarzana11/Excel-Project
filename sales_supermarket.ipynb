```python
best_city = df.groupby("City")["Total"].sum().idxmax()
best_sales = df.groupby("City")["Total"].sum().max()

pd.Series({
    "Best Sales City": best_city,
    "Total Sales": best_sales
})
```

```python
plt.figure(figsize=(7,5))

city_sales = df.groupby("City")["Total"].sum().sort_values(ascending=False)

sns.barplot(
    x=city_sales.index,
    y=city_sales.values,
    palette="pastel",
    hue=city_sales.index,
    legend=False
)

plt.title("Sales by City", fontsize=15, fontweight="bold")
plt.xlabel("City")
plt.ylabel("Total Sales")
plt.show()
```

```python
best_product = (
    df[df["City"] == "Naypyitaw"]
    .groupby("Product line")["Total"]
    .sum()
    .idxmax()
)

best_product
```

```python
naypyitaw_sales = (
    df[df["City"] == "Naypyitaw"]
    .groupby("Product line")["Total"]
    .sum()
    .sort_values(ascending=False)
)

plt.figure(figsize=(10, 6))

sns.barplot(
    x=naypyitaw_sales.index,
    y=naypyitaw_sales.values,
    hue=naypyitaw_sales.index,
    palette="pastel",
    legend=False
)

plt.title("Sales by Product Line in Naypyitaw", fontsize=15, fontweight="bold")
plt.xlabel("Product Line")
plt.ylabel("Total Sales")
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

```python
naypyitaw_sales = df[df["City"] == "Naypyitaw"]["Total"].sum()

plt.figure(figsize=(6, 4))

plt.bar(["Naypyitaw"], [naypyitaw_sales])

plt.title("Total Sales - Naypyitaw", fontsize=15, fontweight="bold")
plt.ylabel("Total Sales")

plt.show()
```
